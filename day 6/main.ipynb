```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return f"{self.name} has deposited {amount}. New balance is {self.balance}"

    def withdraw(self, amount):
        if amount > self.balance:
            return f"{self.name} has insufficient funds. Current balance is {self.balance}"
        else:
            self.balance -= amount
            return f"{self.name} has withdrawn {amount}. New balance is {self.balance}"


# Everything below is NOT indented: it runs after the class is defined
name = input("Enter the name of account holder: ")
balance = int(input("Enter the initial balance: "))

acc1 = BankAccount(name, balance)

deposit_amount = int(input("Enter the amount to deposit: "))
print(acc1.deposit(deposit_amount))

withdraw_amount = int(input("Enter the amount to withdraw: "))
print(acc1.withdraw(withdraw_amount))
```

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        if self.balance < amount:
            print("error: insufficient funds")
        else:
            self.balance += amount
            print(f"{self.name} has deposited {amount}. New balance is {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            return f"{self.name} has insufficient funds. Current balance is {self.balance}"
        else:
            self.balance -= amount
            return f"{self.name} has withdrawn {amount}. New balance is {self.balance}"


# Everything below is NOT indented: it runs after the class is defined
name = input("Enter the name of account holder: ")
balance = int(input("Enter the initial balance: "))

acc1 = BankAccount(name, balance)

deposit_amount = int(input("Enter the amount to deposit: "))
print(acc1.deposit(deposit_amount))

withdraw_amount = int(input("Enter the amount to withdraw: "))
print(acc1.withdraw(withdraw_amount))
```
